```python
class Car:
    def __init__(self, window, door, enginetype):
        self.windows = window
        self.doors = door
        self.engine_type = enginetype

    def self_driving(self):
        return "This is a {} car".format(self.engine_type)
```

# OOPs - Public, Private, Protected

```python
#### In python all the class preporties are public
class Car():
    def __init__(self, window, door, enginetype):
        self.windows = window
        self.doors = door
        self.engine_type = enginetype
```

```python
audi = Car(4, 5, "Diesel")
```

```python
dir(audi)
```

```python
#### Here all the class preporties are protected
class Car():
    def __init__(self, window, door, enginetype):
        self._windows = window
        self._doors = door
        self._engine_type = enginetype
```

```python
audi = Car(4,4,"Petrol")
dir(audi)
```

```python
class Truck(Car):
    def __init__(self, windows, doors, enginetype, horsepower):
        super().__init__(windows, doors, enginetype)
        self.horsepower = horsepower
```

```python
truck1 = Truck(3, 2, "Diesel", 4000)
dir(truck1)
```

## Private variable

```python
#### Here all the class preporties are private
class Car():
    def __init__(self, window, door, enginetype):
        self.__windows = window
        self.__doors = door
        self.__engine_type = enginetype
```

```python
bmw = Car(4, 3, "Diesel")
dir(bmw)
```

## Python OOPs - Inheritance

```python
class Car():
    def __init__(self, window, door, enginetype):
        self.windows = window
        self.doors = door
        self.engine_type = enginetype
    def drive(self):
        print("The Person drives the car")
```

```python
car = Car(4, 5, "petrol")
```

```python
car.windows = 3
print(car.windows)
```

```python
dir(car)
```

```python
car.drive()
```

```python
class audi(Car):
    def __init__(self, window, door, enginetype, enableai):
        super().__init__(window, door, enginetype)
        self.enableai = enableai
    def selfdriving(self):
        print('This car supports Self driving')
```

```python
car100 = audi(3, 2, "Gas", True)
dir(car100)
```

```python
car100.drive()
```

```python
car100.selfdriving()
```

```python
audi_q7 = audi(5, 5, 'Gas', True)
```

```python
dir(audi_q7)
```
