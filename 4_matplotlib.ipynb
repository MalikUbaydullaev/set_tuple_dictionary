## Matplotlib

```python
# import of libraries
import matplotlib.pyplot as plt

import numpy as np

%matplotlib inline
```

```python
# Simple examples

x = np.arange(0,10)
y = np.arange(11,21)
```

```python
a = np.arange(40,50)
b = np.arange(50,60)
```

```python
## plt scatter

plt.scatter(x,y,c = 'g')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')
plt.savefig('Test.png')
```

```python
y = x*x
```

```python
## plt.plot
plt.plot(x, y, 'r*--', linewidth = 2, markersize = 12)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph')
```

```python
# Creating Subplots

plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(x,y,'g--')
plt.subplot(2,2,3)
plt.plot(x,y,'b*-')
plt.subplot(2,2,4)
plt.plot(x,y,'yo')
```

```python
x = np.arange(1,11)
y = 3 * x + 5
plt.title('Matplotlib')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x,y)
plt.show()
```

```python
# Compute the x and y coordinates for points on a sine curve

x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)

plt.title('Sine wave form')
plt.xlabel('X axis')
plt.ylabel('Y axis')
#Plot the points using matplotlib
plt.plot(x, y)
plt.show()
```

```python
# Compute the x and y coordinates for points on a cos curve

x = np.arange(0, 4 * np.pi, 0.1)
y = np.cos(x)

plt.title('Cos wave form')
plt.xlabel('X axis')
plt.ylabel('Y axis')
#Plot the points using matplotlib
plt.plot(x, y)
plt.show()
```

```python
x = np.arange(0, 4 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# First plot
plt.subplot(2, 1, 1)
plt.title('Sin')
plt.plot(x, y_sin)

#Second plot
plt.subplot(2, 1, 2)
plt.xlabel('Cos')
plt.plot(x, y_cos, 'g')

# figure show
plt.show()
```

## Bar plot

```python
# Bar plot

x = [2, 8, 10]
y = [11, 16, 9]

x2 = [3, 9, 11]
y2 = [6, 15, 7]

plt.bar(x,y)
plt.bar(x2, y2, color = 'green')
plt.title('Bar graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

## Histogram

```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a, bins = 20)
plt.title('Histogram')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Boxplot

```python
data = [np.random.normal(0, std, 100) for std in range (1, 4)]

#boxplots
plt.boxplot(data, orientation = 'vertical', patch_artist = True)
plt.title('Boxplot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

## Pie Chart

```python
labels = 'Python', 'Java', 'C++', 'Ruby'
sizes = [215, 130, 245, 210]
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0.1, 0, 0, 0) # explode first slice

#plot
plt.pie(sizes, explode = explode, labels = labels, colors = colors, autopct = '%1.1f%%', shadow = True)

plt.axis('equal')
plt.show()
```
