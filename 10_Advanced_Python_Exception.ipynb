## Python Exception Handling

```python
try:
    ## code block where exception can occur
    a = b
except:
    ## print what kinf of problem have occured  
    print("Some problem may have occured")
```

```python
try:
    a = b
except Exception as ex:
    print(ex)
```

```python
try:
    a = b
except NameError as ex1:
    print("the user have not defined the variable")
except Exception as ex:
    print(ex)
```

```python
# Exception srabatyvayet pi posledovatelnosti
try:
    a = 1
    b = 's'
    c = a + b
except NameError as ex1:
    print("the user have not defined the variable")
except Exception as ex:
    print(ex)
```

```python
# Exception srabatyvayet pi posledovatelnosti
try:
    a = 1
    b = 's'
    c = a + b
except NameError:
    print("the user have not defined the variable")
except TypeError:
    print("Type Error oshibka tipa")
except Exception as ex:
    print(ex)
```
