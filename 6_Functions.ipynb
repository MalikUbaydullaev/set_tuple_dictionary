## Functions in Python

```python
num = 24

if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def even_odd(num):
    if num % 2 == 0:
        print("The number is even")
    else:
       print("The number is odd")
```

```python
def hello(name, age = 29):
    print(f"My name is {name} and age is {age}")
```

```python
hello("Malik", 40)
```

```python
def hello2(name, age = 30):
    print("My name is {} and age is {}".format(name, age))
```

```python
hello2('Salomat', 25)
```

```python
def hello3(*args, **kwargs):
    print(args)
    print(kwargs)
```

```python
hello3("Malik", "Ubaydullo", age = 40, bob = 1986)
```

```python
lst = ["Malik", "Ubaydullo"]
dict_args = {'age': 40, 'bob': 1986}
```

```python
hello3(lst, dict_args)
```

```python
lst2 = [1,2,3,4,5,6,7]
```

```python
def even_odd_sum(lst):
    even_sum = 0
    odd_sum = 0
    for i in lst:
        if i % 2 == 0:
            even_sum = even_sum + i
        else:
            odd_sum = odd_sum + i
    return even_sum, odd_sum
```

```python
even_odd_sum(lst2)
```
